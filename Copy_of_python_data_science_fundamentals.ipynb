```python
# Day 1 - Python Basics

name = "Durai"
age = 20
course = "Cyber Security"

print("Name:", name)
print("Age:", age)
print("Course:", course)

# Data Types
integer_value = 25
float_value = 10.5
string_value = "Python"
boolean_value = True

print(type(integer_value))
print(type(float_value))
print(type(string_value))
print(type(boolean_value))

# Arithmetic Operators
a = 20
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python

marks = 75

if marks >= 50:
    print("Result: Pass")
else:
    print("Result: Fail")

# For loop
print("For Loop:")
for i in range(1, 6):
    print(i)

# While loop
print("While Loop:")
count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
# Day 3 - Lists, Dictionaries and Functions

# List
skills = ["Python", "Cyber Security", "Networking", "Data Science"]

print("Skills:")
for skill in skills:
    print(skill)

# Dictionary
student = {
    "name": "Durai",
    "department": "Cyber Security",
    "year": 2
}

print("\nStudent Details:")
print("Name:", student["name"])
print("Department:", student["department"])
print("Year:", student["year"])

# Function
def calculate_sum(a, b):
    return a + b

result = calculate_sum(10, 20)

print("\nFunction Result:", result)
```

```python
# Basic Data Science Example

import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [75, 82, 68, 90, 85]
}

df = pd.DataFrame(data)

print("Student Marks Data:")
display(df)

print("\nAverage Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
```
